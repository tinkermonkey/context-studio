```python
# Discover and print all of the methods of the pizza_parser object
for method in dir(pizza_parser):
    if callable(getattr(pizza_parser, method)) and not method.startswith('__'):
        print(f"Method: {method}")
        try:
            result = getattr(pizza_parser, method)()
            print(f"Result: {result}")
        except Exception as e:
            print(f"Error calling method {method}: {e}")
```

# py_ez_wikidata

```python
!{sys.executable} -m pip install py-ez-wikidata
```

```python
from ez_wikidata.wdsearch import WikidataSearch

wd_search = WikidataSearch()
wd_search.language = 'en'
results = wd_search.search('Truck', limit = 5)
for result in results:
    print(f"ID: {result['id']}, Label: {result['label']}, Description: {result['description']}")

# pick the first result and get its details
if results:
    first_result = results[0]
    
    # Print all attributes of the wd_search object
    for attr in dir(first_result):
        if not attr.startswith('__'):
            print(f"{attr}: {getattr(first_result, attr)}")

    # Print all of the keys and values of the first_result dictionary
    for key, value in first_result.items():
        print(f"{key}: {value}")
```

```python
# print all of the attributes of the query object
for attr in dir(query):
    if not attr.startswith('__'):
        print(f"{attr}: {getattr(query, attr)}")

print(query.item())
```

# wikidata-client

```python
!{sys.executable} -m pip install wikidata-client
```

```python
import wikidata_client

# query for items that are instance of Q43193 (truck)
results = wikidata_client.query("""\
   SELECT ?item ?itemLabel
   WHERE {
     ?item wdt:P31 wd:Q43193.
     SERVICE wikibase:label { bd:serviceParam wikibase:language "en". }
   }
   LIMIT 5
""")

subclass_results = wikidata_client.query("""
  SELECT ?subclass ?subclassLabel
  WHERE {
    wd:Q43193 wdt:P279 ?subclass.
    SERVICE wikibase:label { bd:serviceParam wikibase:language "en". }
  }
""")
for res in subclass_results:
  print(res)
```
