## Test it out

```python
import spacy
import textdescriptives as td

test_sentance = "Apple is looking at buying U.K. startup for $1 billion"
test_definition = "A medical doctor specializing in diagnosing, treating, and preventing mental, emotional, and behavioral disorders through various therapeutic approaches, including psychotherapy and medication management in children."
test_complication = "I hate tacos. Those things are awful."

nlp = spacy.load("en_core_web_sm")
nlp.add_pipe("textdescriptives/all")

doc = nlp(test_definition)
for token in doc:
    #print(token.text, token.pos_, token.dep_)
    print(token.text, token.lemma_, token.pos_, token.tag_, token.dep_,
            token.shape_, token.is_alpha, token.is_stop)
```
